## Classification


### Introduction

This chapter introduces **classification**, a supervised learning task where the goal is to assign each instance to one of two or more discrete categories. Unlike regression, which predicts continuous values, classification predicts class labels. A central example in this chapter is the **MNIST dataset**, a collection of handwritten digit images widely used to benchmark image classification algorithms.

Classification problems are ubiquitous: spam detection, disease diagnosis, credit approval, and image recognition all involve deciding which class an instance belongs to.

### The MNIST Dataset

The MNIST dataset contains **70,000 grayscale images** of handwritten digits (0–9), each **28×28 pixels** in size. Every image is labeled with the correct digit. The dataset is split into **60,000 training** and **10,000 test** examples.

Each image can be represented as a 784-dimensional vector (flattening 28×28 pixels). The dataset is commonly used for experimenting with classification algorithms and data preprocessing techniques.

Visualizing the data helps build intuition about class boundaries and data variability:

### Training a Binary Classifier

A **binary classifier** distinguishes between two possible classes. As an example, the chapter builds a system to detect whether an image represents the digit '5' or not ('5-detector').

Steps:
1. Create target vectors: `y_train_5` and `y_test_5`, containing `True` for digit 5 and `False` otherwise.
2. Train a simple linear model using stochastic gradient descent (`SGDClassifier`).
3. Evaluate predictions using suitable performance metrics.

Binary classification serves as the foundation for understanding metrics like precision and recall.

<p align="left"><img src="../fig/figure3.3.png" width="45%"></p>

### Performance Measures

Using accuracy alone is insufficient, especially in **imbalanced datasets**. For instance, if only 10% of images are '5's, a model predicting 'not 5' always achieves 90% accuracy but is useless. Thus, other metrics are used:

- **Confusion Matrix:** shows true vs. predicted classifications.
- **Precision:** fraction of positive predictions that are correct.
- **Recall (Sensitivity):** fraction of actual positives correctly identified.
- **F1 Score:** harmonic mean of precision and recall.

These metrics together provide a comprehensive view of model performance.

<p align="left"><img src="../fig/figure3.4.png" width="45%"></p>

### Precision and Recall Trade-Off

Classification models output **decision scores** that can be thresholded to classify an instance as positive or negative. By varying this threshold, we can adjust the trade-off between precision and recall:

- **Higher threshold:** Model is more conservative — precision increases, recall decreases.
- **Lower threshold:** Model detects more positives — recall increases, precision decreases.

Plotting **precision-recall curves** helps visualize this relationship and identify the optimal threshold.

<p align="left"><img src="../fig/figure3.5.png" width="45%"></p>

### The ROC Curve

Another important diagnostic tool is the **Receiver Operating Characteristic (ROC) curve**, which plots the **True Positive Rate (Recall)** versus the **False Positive Rate (1 - Specificity)** for different threshold settings.

- A perfect classifier’s curve reaches the top-left corner.
- The **Area Under the Curve (AUC)** measures overall performance (1.0 = perfect, 0.5 = random).
- Use **precision-recall curves** instead of ROC when dealing with highly imbalanced classes.

### Multiclass Classification

Most real-world problems involve **more than two classes**. Some algorithms (e.g., Random Forest, Neural Networks) handle multiclass classification natively, while others require strategies like:

- **One-vs-All (OvA):** trains one classifier per class, distinguishing that class from all others.
- **One-vs-One (OvO):** trains one classifier for every pair of classes.

Scikit-Learn automatically applies these strategies when necessary.

<p align="left"><img src="../fig/figure3.7.png" width="45%"></p>

### Error Analysis

After training a classifier, performing **error analysis** is critical for improvement:

- Inspect the **confusion matrix** to identify which classes are most often confused.
- Visualize misclassified examples to spot ambiguous or mislabeled instances.
- Examine feature importance or use model interpretability tools to understand mistakes.

Such analysis informs better preprocessing, feature engineering, or data collection strategies.

### Multilabel Classification

In **multilabel classification**, each instance can belong to multiple classes simultaneously. For example, a photo-tagging system might output multiple labels such as `['person', 'outdoor', 'smiling']`.

Evaluation metrics include:
- **Macro-averaged F1 score:** averages F1 across labels.
- **Subset accuracy:** fraction of instances with all labels predicted correctly.

Multilabel learning enables models to capture richer relationships between labels.

### Multioutput Classification

A **multioutput (or multitarget)** classifier predicts multiple outputs per instance. Each target can have multiple possible values. This setup generalizes both multiclass and multilabel tasks.

Example: A system that takes a noisy image of a handwritten digit and outputs a clean version (predicting each pixel intensity) performs multioutput regression/classification.

```python
# Example: Binary classification using SGDClassifier on MNIST
from sklearn.datasets import fetch_openml
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
import numpy as np

mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X, y = mnist.data, mnist.target.astype(np.uint8)
X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]

y_train_5 = (y_train == 5)
y_test_5 = (y_test == 5)

sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train, y_train_5)

y_train_pred = cross_val_predict(sgd_clf, X_train, y_train_5, cv=3)

cm = confusion_matrix(y_train_5, y_train_pred)
prec = precision_score(y_train_5, y_train_pred)
rec = recall_score(y_train_5, y_train_pred)
f1 = f1_score(y_train_5, y_train_pred)

print('Confusion matrix:\n', cm)
print('Precision:', prec)
print('Recall:', rec)
print('F1 Score:', f1)
```

### Summary and Best Practices

- Use **precision, recall, and F1** instead of accuracy for imbalanced classes.
- Plot **precision-recall** and **ROC curves** to select optimal thresholds.
- Apply **cross-validation** to evaluate generalization.
- Use **OvA** or **OvO** strategies for multiclass tasks.
- Perform **error analysis** to guide improvements.
- For multilabel or multioutput problems, adapt metrics accordingly.

These methods together form a strong foundation for building reliable classification systems.

## Training Models


### Introduction

This chapter explores **how models learn from data** by minimizing a cost function using optimization algorithms. The chapter focuses primarily on **Linear Regression**, **Gradient Descent**, and **Regularized models** (Ridge, Lasso, Elastic Net), before extending to **Logistic Regression** and **Softmax Regression** for classification.

Understanding these foundations is essential: even complex neural networks rely on the same underlying principles — parameterized models optimized by gradient-based methods.

### Linear Regression

Linear Regression aims to find a linear relationship between input features and the target variable:

$$ \hat{y} = \theta_0 + \theta_1 x_1 + \dots + \theta_n x_n = X \theta $$

where $X$ is the design matrix of shape $(m, n+1)$ and $\theta$ is the parameter vector (including bias). The objective is to minimize the **Mean Squared Error (MSE)** cost:

$$ J(\theta) = \frac{1}{m} \sum_{i=1}^{m} (\hat{y}^{(i)} - y^{(i)})^2 = \frac{1}{m} (X\theta - y)^T (X\theta - y) $$

Geometrically, the model fits a hyperplane that minimizes the squared distance between predictions and actual values.

<p align="left"><img src="../fig/figure4.2.png" width="45%"></p>

### The Normal Equation

The Normal Equation provides a **closed-form analytical solution** for the optimal parameters:

$$ \hat{\theta} = (X^T X)^{-1} X^T y $$

It directly computes the weights that minimize the MSE. However:
- Complexity is $O(n^3)$ due to matrix inversion — inefficient for large feature sets.
- Requires $(X^T X)$ to be invertible (no redundant or correlated features).

Despite being exact, it is rarely used for high-dimensional data. Instead, **iterative methods** like Gradient Descent are preferred for scalability.

<p align="left"><img src="../fig/figure4.3.png" width="45%"></p>

### Gradient Descent — The Core of Learning

Gradient Descent is a fundamental iterative optimization technique used to minimize a cost function by moving the parameters in the direction of the **negative gradient**:

$$ \theta := \theta - \eta \nabla_\theta J(\theta) $$

where $\eta$ is the **learning rate**. The gradient represents the slope of the cost surface; subtracting it moves parameters downhill toward a minimum.

Key aspects:
- A **small learning rate** ensures stability but converges slowly.
- A **large learning rate** risks overshooting or divergence.
- Scaling input features (standardization) is crucial for efficient convergence.

<p align="left"><img src="../fig/figure4.4.png" width="45%"></p>

### Types of Gradient Descent

1. **Batch Gradient Descent:**
   - Computes gradient using the **entire training set**.
   - Guarantees convergence to the global minimum for convex cost functions.
   - Expensive for large datasets.

2. **Stochastic Gradient Descent (SGD):**
   - Updates parameters using **one random instance** at a time.
   - Introduces randomness → allows escaping local minima.
   - High variance → oscillates near the minimum.

3. **Mini-Batch Gradient Descent:**
   - Processes small random subsets (mini-batches).
   - Combines efficiency of SGD and stability of batch updates.

Modern deep learning relies almost exclusively on **mini-batch gradient descent** with adaptive learning rates (e.g., Adam, RMSProp).

### Polynomial Regression

To capture nonlinear relationships, we can augment features with polynomial terms. For example, transforming $x$ to $(x, x^2, x^3, ...)$ allows a linear model to fit curved patterns.

However, increasing polynomial degree raises risk of **overfitting**. Visualizing fitted curves shows how higher-degree polynomials can perfectly fit training data yet generalize poorly.

<p align="left"><img src="../fig/figure4.6.png" width="45%"></p>

### Learning Curves and the Bias–Variance Tradeoff

Learning curves plot **training error** and **validation error** versus training set size. They reveal the model’s generalization behavior:

- **High bias (underfitting):** both training and validation errors are high.
- **High variance (overfitting):** training error is low, validation error is high.
- **Good fit:** both errors converge to a low value.

These curves guide whether to add features, increase regularization, or gather more data.

### Regularized Linear Models

Regularization discourages large weights to reduce overfitting.

1. **Ridge Regression (L2 Regularization):**
   Adds a penalty on the squared magnitude of coefficients:
   $$ J(\theta) = MSE(\theta) + \alpha \sum_{i=1}^n \theta_i^2 $$
   Encourages small but nonzero coefficients.

2. **Lasso Regression (L1 Regularization):**
   Adds a penalty on the absolute value of coefficients:
   $$ J(\theta) = MSE(\theta) + \alpha \sum_{i=1}^n |\theta_i| $$
   Tends to zero out some weights entirely — performs **feature selection**.

3. **Elastic Net:**
   Combines L1 and L2 penalties to balance sparsity and stability.

### Logistic Regression

For classification tasks, **Logistic Regression** models the probability of belonging to a positive class using the **sigmoid (logistic) function**:

$$ h_\theta(x) = \frac{1}{1 + e^{-\theta^T x}} $$

Decision rule:
- Predict class 1 if $h_\theta(x) \geq 0.5$.
- Predict class 0 otherwise.

Training minimizes the **log loss (cross-entropy)** cost:

$$ J(\theta) = -\frac{1}{m} \sum [y \log(h_\theta(x)) + (1 - y) \log(1 - h_\theta(x))] $$

Logistic Regression outputs interpretable probabilities and can be extended to multiple classes via **Softmax Regression**.

<p align="left"><img src="../fig/figure4.9.png" width="45%"></p>

### Softmax Regression

Softmax Regression generalizes Logistic Regression for multiclass problems by computing the probability of each class:

$$ P(y = k | x) = \frac{e^{s_k(x)}}{\sum_{j=1}^{K} e^{s_j(x)}} $$

where $s_k(x) = \theta_k^T x$. The model assigns the class with the highest probability:

$$ \hat{y} = \arg\max_k P(y = k | x) $$

The cost function is the **cross-entropy loss** across all classes. Softmax regression is the foundation of output layers in neural networks.

<p align="left"><img src="../fig/figure4.10.png" width="45%"></p>

```python
# Demonstration: Gradient Descent with Linear Regression
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

X_b = np.c_[np.ones((100, 1)), X]
eta = 0.1  # learning rate
n_iterations = 1000
m = 100
theta = np.random.randn(2, 1)

for iteration in range(n_iterations):
    gradients = 2/m * X_b.T.dot(X_b.dot(theta) - y)
    theta = theta - eta * gradients

print('Estimated parameters:', theta)

plt.plot(X, y, 'b.')
plt.plot(X, X_b.dot(theta), 'r-', linewidth=2)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent Linear Fit')
plt.show()
```

### Summary and Best Practices

- Linear models form the core of statistical learning; they are interpretable and efficient.
- Gradient Descent is universal — used in regression, classification, and deep learning.
- Feature scaling is essential for fast convergence.
- Regularization (Ridge, Lasso, Elastic Net) balances bias and variance.
- Logistic and Softmax Regression extend linear models to classification tasks.
- Learning curves reveal underfitting vs. overfitting.

These principles underpin nearly all modern machine learning algorithms.

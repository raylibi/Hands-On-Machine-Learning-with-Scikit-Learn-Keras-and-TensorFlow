## The Machine Learning Landscape


### What Is Machine Learning?

Machine Learning (ML) is the scientific discipline and engineering practice of building systems that improve their performance at tasks through experience (data) rather than explicit programming. Two classic definitions help clarify the idea:

> **Arthur Samuel (1959):** "Machine Learning is the field of study that gives computers the ability to learn without being explicitly programmed."

> **Tom Mitchell (1997):** "A computer program is said to learn from experience E with respect to some class of tasks T and performance measure P, if its performance at tasks in T, as measured by P, improves with experience E."

A simple running example used throughout this chapter is a **spam filter**:
- **Task (T):** classify incoming email as spam or not spam.
- **Experience (E):** a dataset of labeled emails (spam/ham).
- **Performance (P):** accuracy, precision/recall, or F1 score on new emails.

If the filter gets better as it sees more labeled examples, it is learning from experience.

### Why Use Machine Learning?

Traditional programming requires human experts to encode rules. For many modern problems this is impractical because:
1. The number of rules would be enormous and brittle.
2. The environment changes quickly (spam tactics, market behavior, user tastes).
3. The underlying patterns are high-dimensional and hard to express by hand.

Machine Learning addresses these points by automatically **discovering statistical patterns** in data and adapting as data changes. Key motivations include:

- **Scalability & Complexity:** ML handles complex mappings between inputs and outputs when handcrafted rules fail.
- **Adaptability:** When new data arrives, models can be retrained or updated incrementally to reflect changes.
- **Feature extraction:** Modern models (especially deep learning) can learn useful features directly from raw data (images, audio, text).
- **Insight discovery (data mining):** Trained models and analyses can reveal which factors influence predictions most, yielding actionable insights.

### Examples of Machine Learning Applications

- **Image classification & detection:** quality control on production lines, face recognition, medical imaging (tumor detection).
- **NLP tasks:** text classification, sentiment analysis, machine translation, summarization.
- **Speech recognition:** converting audio to text.
- **Forecasting & regression:** sales forecasting, demand prediction, pricing.
- **Anomaly detection:** fraud detection, predictive maintenance.
- **Clustering & segmentation:** customer segmentation for marketing.
- **Reinforcement learning:** training agents to act optimally in environments (robots, games).

These examples illustrate ML's breadth — from narrow predictive tasks to complex decision-making agents.

### Types of Machine Learning Systems

We can classify ML systems in multiple ways: by supervision, by learning mode (batch vs online), or by generalization approach (instance-based vs model-based).

#### Supervised Learning

Supervised learning trains models on labeled data (input-output pairs). Two main supervised tasks:
- **Classification:** predict discrete labels (e.g., spam vs ham).
- **Regression:** predict continuous values (e.g., house price).

Common algorithms: Linear Regression, Logistic Regression, k-NN, SVM, Decision Trees, Random Forests, and Neural Networks.

#### Unsupervised Learning

Unsupervised learning uses unlabeled data to discover structure: clustering, dimensionality reduction, association rules, and anomaly detection.

Typical tools: K-Means, DBSCAN, Hierarchical Clustering, PCA, t-SNE, Isolation Forest.

#### Semisupervised Learning

When labeled data is scarce and unlabeled data plentiful, semisupervised methods use both to improve performance. For instance, clustering can group similar inputs, and a few labels can then propagate to clusters.

#### Reinforcement Learning

An agent learns by interacting with an environment and receiving rewards/penalties. The goal is to learn a policy (mapping states to actions) that maximizes cumulative reward. Classic successes include AlphaGo and many robotics/control tasks.

### Batch and Online Learning

In **batch learning**, models are trained on the full dataset, then deployed as fixed models until retrained. In **online learning**, models update incrementally as new data arrives, enabling adaptation to streams and very large datasets via mini-batches.

<p align="left"><img src="../fig/figure1.13.png" width="45%"></p>
<p align="left"><img src="../fig/figure1.14.png" width="45%"></p>

Online learning is powerful for dynamic environments but risky if corrupted or adversarial data enters the stream; monitoring and rollback mechanisms are essential.

### Instance-Based vs Model-Based Learning

- **Instance-based learning** stores training instances and makes predictions by comparing new instances to stored ones (e.g., k-NN).
- **Model-based learning** fits a parametric or nonparametric model (e.g., linear model, neural network) and uses it for predictions.

### Main Challenges of Machine Learning

Two broad failure modes exist: *bad data* and *bad algorithm/architecture/hyperparameters*. The main data-related challenges include:

#### Insufficient Quantity of Training Data
Many ML methods require large datasets. For hard tasks like image or speech recognition, millions of labeled examples may be necessary unless transfer learning or pretrained components are used.

#### Nonrepresentative Training Data
Training data must match the distribution of examples the system will see in production. Sampling bias or dataset shift leads to poor real-world performance.

#### Poor-Quality Data
Missing values, incorrect labels, and noisy measurements all degrade models. Data cleaning is often the most time-consuming part of ML work.

#### Irrelevant Features
Including irrelevant or redundant features hurts learning. Feature selection, engineering, and dimensionality reduction are vital.

#### Overfitting
A model that fits the training set too closely (including noise) will perform poorly on unseen data. Regularization, simpler models, more data, and cross-validation help mitigate overfitting.

#### Underfitting
A model that is too simple will not capture underlying patterns. Solutions include using more powerful models, adding features, or reducing regularization.

### Testing and Validating

To estimate how well a model generalizes, split data into training and test sets. Train on the training set, tune on a validation set (or use cross-validation), and report final performance on the test set. Beware of leaking information between sets.

If training error is low but test error is high, you are likely overfitting. Use learning curves to diagnose whether more data or a simpler/more complex model is required.

### Hyperparameter Tuning and Model Selection

Hyperparameters (e.g., regularization strength, number of neighbors in k-NN, tree depth) should be tuned using a validation set or cross-validation. Avoid tuning on the test set to prevent overfitting to the test set itself. Common strategies include grid search and randomized search.

After selecting hyperparameters, retrain the final model on the entire training set (training + validation) and evaluate once on the held-out test set.

### Data Mismatch

Sometimes large datasets are available but differ from the production distribution (e.g., web images vs. mobile app photos). In this case, create a representative validation/test set (from production-like data) and consider domain adaptation or preprocessing to align datasets.

Andrew Ng's suggested diagnostic: create a 'train-dev' set from the large but mismatched data and compare performance to the true validation set to determine whether the problem is overfitting or data mismatch.

### No Free Lunch Theorem

No single learning algorithm works best for every problem. The No Free Lunch theorem formalizes this: without prior assumptions about the problem, no algorithm is universally superior. In practice, choose reasonable model families based on domain knowledge and validate empirically.

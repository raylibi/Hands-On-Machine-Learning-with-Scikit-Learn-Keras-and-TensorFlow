## Support Vector Machines


### Introduction

Support Vector Machines (SVMs) are one of the most powerful and versatile supervised learning algorithms. They are capable of performing **linear or nonlinear classification**, regression, and even outlier detection.

The core idea of SVMs is to find a decision boundary that separates classes with the **largest possible margin**, which improves generalization on unseen data. SVMs are conceptually elegant, geometrically interpretable, and often effective in high-dimensional spaces.

<p align="left"><img src="../fig/figure5.1.png" width="45%"></p>

### Linear SVM Classification

A **Linear SVM Classifier** aims to find the optimal separating hyperplane:

$$ \mathbf{w}^T \mathbf{x} + b = 0 $$

where $\mathbf{w}$ is the weight vector and $b$ is the bias. The goal is to maximize the **margin**, the distance between the decision boundary and the nearest training points (called **support vectors**).

The margin width is given by:

$$ \text{Margin} = \frac{2}{\|\mathbf{w}\|} $$

The optimization objective becomes:

$$ \min_{\mathbf{w},b} \frac{1}{2}\|\mathbf{w}\|^2 $$

subject to $y_i(\mathbf{w}^T\mathbf{x}_i + b) \geq 1$ for all $i$.

This is a convex quadratic optimization problem, ensuring a unique global minimum.

<p align="left"><img src="../fig/figure5.2.png" width="45%"></p>

### Soft Margin Classification

Real-world data is rarely perfectly separable. The **Soft Margin SVM** introduces slack variables $\xi_i$ to allow some misclassifications:

$$ \min_{\mathbf{w},b} \frac{1}{2}\|\mathbf{w}\|^2 + C \sum_i \xi_i $$

subject to $y_i(\mathbf{w}^T\mathbf{x}_i + b) \geq 1 - \xi_i$, and $\xi_i \geq 0$.

The hyperparameter **C** controls the trade-off:
- **Large C** → narrow margin, fewer violations (risk of overfitting).
- **Small C** → wider margin, more violations (risk of underfitting).

This mechanism makes SVMs robust to noisy data.

<p align="left"><img src="../fig/figure5.3.png" width="45%"></p>

### Nonlinear SVM Classification

Linear decision boundaries are often insufficient. To handle nonlinear relationships, we can use **feature mappings** that project inputs into higher-dimensional spaces.

For example, mapping $(x_1, x_2)$ to $(x_1, x_2, x_1^2, x_2^2, x_1x_2)$ allows a linear model to capture nonlinear patterns. However, explicitly computing these transformations is costly.

This motivates the **kernel trick**, which allows SVMs to compute inner products in high-dimensional feature spaces implicitly.

<p align="left"><img src="../fig/figure5.4.png" width="45%"></p>

### The Kernel Trick

A **kernel function** computes the similarity between two input vectors $x$ and $x'$ without explicitly transforming them:

$$ K(x, x') = \phi(x)^T \phi(x') $$

Common kernels:
- **Polynomial kernel:** $K(x, x') = (x^T x' + 1)^d$
- **Radial Basis Function (RBF):** $K(x, x') = \exp(-\gamma \|x - x'\|^2)$
- **Sigmoid kernel:** $K(x, x') = \tanh(\kappa x^T x' + c)$

Kernelized SVMs can model complex boundaries efficiently.

<p align="left"><img src="../fig/figure5.5.png" width="45%"></p>

### Polynomial and RBF Kernels

**Polynomial kernels** allow curved decision boundaries of varying flexibility based on degree *d*. Higher degrees model more complex patterns but may overfit.

The **RBF kernel**, also known as the **Gaussian kernel**, measures similarity by distance — nearby points have high kernel values. The parameter $\gamma$ controls the width:
- **Small $\gamma$** → broader influence, smoother decision boundary.
- **Large $\gamma$** → tighter influence, risk of overfitting.

<p align="left"><img src="../fig/figure5.6.png" width="45%"></p>

### SVM Decision Function Geometry

The SVM’s prediction depends only on **support vectors**, the subset of training points lying on or within the margin boundaries. This makes SVMs **sparse** and efficient at inference time.

The decision function can be written as:

$$ f(x) = \sum_i \alpha_i y_i K(x_i, x) + b $$

where $\alpha_i$ are the Lagrange multipliers from the dual optimization problem. Only nonzero $\alpha_i$ correspond to support vectors.

<p align="left"><img src="../fig/figure5.7.png" width="45%"></p>

### SVM Regression (SVR)

SVMs can also perform regression using the **ε-insensitive loss**. The objective is to fit as many instances as possible within a tube of radius ε around the regression line:

$$ |y - f(x)| \leq \varepsilon $$

Errors outside the tube are penalized. The regularization parameter **C** controls how much the model tolerates deviations:
- High C → narrower tube, risk of overfitting.
- Low C → wider tube, risk of underfitting.

<p align="left"><img src="../fig/figure5.8.png" width="45%"></p>

### Computational Complexity and Scalability

SVM training involves solving a quadratic optimization problem with complexity between $O(m^2)$ and $O(m^3)$, where *m* is the number of training instances.

For very large datasets, this becomes computationally expensive. Practical solutions include:
- Using **linear SVMs** with libraries like `LinearSVC`.
- Using **SGDClassifier** with hinge loss for approximate linear SVMs.
- Employing **kernel approximations** (e.g., Random Fourier Features).

<p align="left"><img src="../fig/figure5.9.png" width="45%"></p>

```python
# Example: Linear vs RBF Kernel SVM
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
import numpy as np
import matplotlib.pyplot as plt

# Load a simple dataset
X, y = datasets.make_moons(noise=0.2, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train two models
linear_svm = SVC(kernel='linear', C=1)
rbf_svm = SVC(kernel='rbf', gamma=1, C=1)

linear_svm.fit(X_train_scaled, y_train)
rbf_svm.fit(X_train_scaled, y_train)

print('Linear SVM accuracy:', linear_svm.score(X_test_scaled, y_test))
print('RBF SVM accuracy:', rbf_svm.score(X_test_scaled, y_test))
```

### Summary and Insights

- SVMs maximize the margin to improve generalization.
- The parameter **C** controls regularization, balancing margin width and misclassification.
- The **kernel trick** allows efficient nonlinear classification.
- **RBF kernels** are widely used for general-purpose nonlinear problems.
- **SVR** applies the same principles to regression with ε-insensitive loss.
- For large datasets, prefer **LinearSVC** or approximate methods.

Support Vector Machines remain a strong baseline algorithm — interpretable, mathematically principled, and robust across many applications.
